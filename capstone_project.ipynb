```python
#   task 5
# Add a new bonus column in the Final dataframe. Give a 5% bonus concerning project
#cost only to employees who have finished the projects.
Final["Bonus"]=0
Final.loc[Final["Status"] == "Finished",
"Bonus"]=Final["Cost"]*0.05
Final
```

```python
#save updated Final DataFrame
Final.to_csv("final.csv",index=False)
```

```python
#   task 6
#Demote employees whose project status is Failed
Final.loc[Final["Status"]=="Failed",
"Designation Level"] +=1
Final
```

```python
# Reload original CSV files
project_df = pd.read_csv("Project.csv")
employee_df = pd.read_csv("Employee.csv")
seniority_df = pd.read_csv("seniority.csv")

# Join them again
Final = pd.merge(project_df, employee_df, on="ID")
Final = pd.merge(Final, seniority_df, on="ID")

# Check shape
Final.shape
```

```python
# Task 6

# Decrease Designation Level by 1 where Status is Failed
Final.loc[Final["Status"] == "Failed", "Designation Level"] = Final.loc[
    Final["Status"] == "Failed", "Designation Level"
] - 1

# Display DataFrame
Final
```

```python
# Save updated Final DataFrame
Final.to_csv("Final.csv", index=False)
```

```python
# Task 7

# Add Mr. for Male employees
# Add Mrs. for Female employees
Final.loc[Final["Gender"] == "M", "First Name"] = "Mr. " + Final["First Name"]
Final.loc[Final["Gender"] == "F", "First Name"] = "Mrs. " + Final["First Name"]

# Remove Gender column
Final.drop("Gender", axis=1, inplace=True)
Final
```

```python
# Save updated Final DataFrame
Final.to_csv("Final.csv", index=False)
```

```python
# Task 8
# Increase Designation Level by 1 if Age > 29
Final.loc[Final["Age"] > 29, "Designation Level"] = Final.loc[
    Final["Age"] > 29, "Designation Level"
] + 1

Final
```

```python
# Save updated Final DataFrame
Final.to_csv("Final.csv", index=False)
```

```python
# Task 9
# Group by ID and First Name, then calculate total project cost
TotalProjCost = Final.groupby(["ID", "First Name"])["Cost"].sum().reset_index()

# Rename the Cost column
TotalProjCost.rename(columns={"Cost": "Total Cost"}, inplace=True)

# Display the result
TotalProjCost
```

```python
# Save TotalProjCost DataFrame
TotalProjCost.to_csv("TotalProjCost.csv", index=False)
```
